In this exercise, we will run local LLM inference. Instead of triggering a remote 
API, you can also download a complete LLM (at least some smaller ones) and run inference locally.

```python
from transformers import pipeline, utils
utils.logging.set_verbosity_error() # hide technical warnings

# Load a small instruction-tuned LLM (about 0.5 billion parameters, ~1GB) - the whole model is downloaded
# automatically from huggingface.co and runs fine on a CPU
generator = pipeline("text-generation", model="Qwen/Qwen2.5-0.5B-Instruct")
```

## Example for generating a text

This model is tiny compared to today's large models, but it runs on your own computer. Like the OpenAI API, chat models expect a list of messages:

```python
messages = [{"role": "user", "content": "Tell me a short story that starts with 'Once upon a time'."}]
generated = generator(messages,
                      max_new_tokens=100,
                      do_sample=True, # sample the next token randomly, otherwise temperature has no effect
                      temperature=0.9)
print(generated[0]["generated_text"][-1]["content"])
```
